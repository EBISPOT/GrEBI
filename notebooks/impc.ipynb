```python
import sys
import os
from py2neo import Graph
from pathlib import Path
from pandas import DataFrame

graph = Graph("bolt://localhost:7687")
```

## Which mouse lines most often co-occur in a publication?

```python

df = DataFrame(graph.run("""
MATCH (allele1:`impc:Allele`)<-[:`impc:alleles`]-(pub:`impc:Publication`)-[:`impc:alleles`]->(allele2:`impc:Allele`)
WHERE elementId(allele1) < elementId(allele2) AND allele1.`impc:symbol`[0] <> allele2.`impc:symbol`[0]
WITH collect(DISTINCT pub.`impc:pmId`[0]) AS pubs, allele1.`impc:symbol`[0] as a1, allele2.`impc:symbol`[0] as a2
RETURN a1, a2, size(pubs) AS npubs, pubs
ORDER BY npubs DESC
LIMIT 10
""").data())

print(df.to_markdown())
```

## Which mouse lines most often co-occur in a publication? What are their respective phenotypes?

```python

df = DataFrame(graph.run("""
MATCH (allele1:`impc:Allele`)<-[:`impc:alleles`]-(pub:`impc:Publication`)-[:`impc:alleles`]->(allele2:`impc:Allele`)
WHERE elementId(allele1) < elementId(allele2)
WITH allele1, allele2, collect(DISTINCT pub) AS common_pubs
WITH
    allele1.`impc:symbol` AS a1,
    allele2.`impc:symbol` AS a2, size(common_pubs) AS num_common_pubs, COLLECT {
        MATCH (allele1)<-[:`impc:mouseAlleleId`]-(assoc1:`impc:GenePhenotypeAssociation`)-[:`biolink:has_phenotype`]-(phenotype1:`ols:Class`)
        RETURN phenotype1.`grebi:name`[0]
    } AS p1,
    COLLECT {
        MATCH (allele2)<-[:`impc:mouseAlleleId`]-(assoc2:`impc:GenePhenotypeAssociation`)-[:`biolink:has_phenotype`]-(phenotype2:`ols:Class`)
        RETURN phenotype2.`grebi:name`[0]
    } AS p2
WHERE size(p1) > 0 AND size(p2) > 0
RETURN num_common_pubs, a1, a2, p1, p2
ORDER BY num_common_pubs DESC
LIMIT 10
""").data())

print(df.to_markdown())
```

## For a gene of interest e.g. ALPK3 - what are the statistically significant phenotypes for the gene?

```python

df = DataFrame(graph.run("""
MATCH (gene:`impc:MouseGene`)-[:`impc:mouseAlleleId`]->(allele:`impc:Allele`)
MATCH (allele)<-[:`impc:mouseAlleleId`]-(assoc:`impc:GenePhenotypeAssociation`)-[:`biolink:has_phenotype`]-(phenotype:`ols:Class`)
WHERE "alpha-kinase 3" IN gene.`grebi:name`
RETURN DISTINCT phenotype.`grebi:name`[0] as phenotype
""").data())

print(df.to_markdown())
```

## For a gene of interest e.g. ALPK3 - what are the genes’s GO terms (process, function, and cellular process) associated with the gene  - report hets and homs?

```python

df = DataFrame(graph.run("""
MATCH (gene:`impc:MouseGene`)-[:`biolink:actively_involved_in`]->(goterm:`ols:Class`)
WHERE "alpha-kinase 3" IN gene.`grebi:name`
RETURN goterm.`grebi:name`[0] as go
""").data())

print(df.to_markdown())
```

## What are the disease associations for IMPC genes from OpenTargets?

### For 1 gene

```python
df = DataFrame(graph.run("""
MATCH (mouse_gene:`impc:MouseGene`)-[:`impc:humanGeneOrthologues`]->(human_gene:`hgnc:Gene`)<-[:`otar:targetId`]-(evidence:`otar:Evidence`)-[:`otar:diseaseId`]->(disease:`ols:Class`)
WHERE "alpha-kinase 3" IN mouse_gene.`grebi:name`
RETURN disease.`grebi:name`[0] as disease, evidence.`otar:score`[0] as score
ORDER BY score DESC
LIMIT 10
""").data())

print(df.to_markdown())
```

### For all genes

```python
df = DataFrame(graph.run("""
MATCH (mouse_gene:`impc:MouseGene`)-[:`impc:humanGeneOrthologues`]->(human_gene:`hgnc:Gene`)<-[:`otar:targetId`]-(evidence:`otar:Evidence`)-[:`otar:diseaseId`]->(disease:`ols:Class`)
RETURN mouse_gene.`grebi:name`[0] as mouse_gene_name, disease.`grebi:name`[0] as disease, evidence.`otar:score`[0] as score
ORDER BY score DESC
""").data())

df.to_csv("impc_disease_associations.csv", index=False)
```

# Which genes have the most and least phenotypes by system?

Would be useful information to get per data release as a series if comparative graphs


```python
df = DataFrame(graph.run("""
MATCH (g:`impc:MouseGene`)-[:`biolink:has_phenotype`]->(phenotype:`ols:Class`)
        -[:`upheno:phenotype_affects_entity`]->(anatomical_entity:`ols:Class`)
        -[:`biolink:part_of`]->(system:`ols:Class`)
        -[:`biolink:broad_match`]->(system_root:`ols:Class`)-[:id]->(id:Id { id: "uberon:0000467" })
RETURN g.`grebi:name`[0] AS gene, system.`grebi:name`[0] AS system, count(phenotype) as n_phenotype
ORDER BY n_phenotype DESC
""").data())

print(df.to_markdown())
```

## Which genes affect only one system?

This could be represented as a bipartite graph with genes linking (systems) tissues that are perturbed.
Won’t be tissues unless we layer on additional information e.g. on expression in tissues or cells


```python
df = DataFrame(graph.run("""
MATCH (g:`impc:MouseGene`)-[:`biolink:has_phenotype`]->(phenotype:`ols:Class`)
        -[:`upheno:phenotype_affects_entity`]->(anatomical_entity:`ols:Class`)
        -[:`biolink:part_of`]->(system:`ols:Class`)
        -[:`biolink:broad_match`]->(system_root:`ols:Class`)-[:id]->(id:Id { id: "uberon:0000467" })
WITH collect(system) AS systems, g
WHERE size(systems)=1
RETURN g.`grebi:name`[0] AS gene, systems[0].`grebi:name`[0] AS system
""").data())

print(df.to_markdown())
```

## For all eye phenotypes retrieve the MP terms and associated genes

add GO terms are associated with eye phenotypes 

```python
df = DataFrame(graph.run("""
MATCH (g:`impc:MouseGene`)-[:`biolink:has_phenotype`]->(phenotype:`ols:Class`)-[:`upheno:phenotype_affects_entity`]->(anatomical_entity:`ols:Class`)
WHERE "eye" IN anatomical_entity.`grebi:name`
RETURN [id in phenotype.id WHERE id =~ "mp:[0-9]*" | id][0] AS mp_term,
        phenotype.`grebi:name`[0] AS mp_term_label,
        g.`grebi:name`[0] AS gene
""").data())

print(df.to_markdown())
```
